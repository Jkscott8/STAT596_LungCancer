```python
import pandas as pd

files = [
    ("2018 Lung Cancer Death Rates_reformatted.csv", 2018),
    ("2019 Lung Cancer Death Rates_reformatted.csv", 2019),
    ("2020 Lung Cancer Death Rates_reformatted.csv", 2020),
    ("2021 Lung Cancer Death Rates_reformatted.csv", 2021),
    ("2022 Lung Cancer Death Rates_reformatted.csv", 2022),
    ("2023 Lung Cancer Death Rates_reformatted.csv", 2023),
]

dfs = []

for file, year in files:
    df = pd.read_csv(file)
    df["Year"] = year
    dfs.append(df)

combined = pd.concat(dfs, ignore_index=True)

long_df = combined.melt(
    id_vars=["County", "County Code", "Year"],
    value_vars=[
        "January","February","March","April","May","June",
        "July","August","September","October","November","December"
    ],
    var_name="Month",
    value_name="Death Rate"
)

long_df["Month_Year"] = long_df["Month"] + " " + long_df["Year"].astype(str)

wide_df = long_df.pivot_table(
    index=["County", "County Code"],
    columns="Month_Year",
    values="Death Rate",
    aggfunc="first"
).reset_index()

months = ["January","February","March","April","May","June",
          "July","August","September","October","November","December"]

ordered_cols = ["County", "County Code"]
for year in range(2018, 2024):
    for m in months:
        col = f"{m} {year}"
        if col in wide_df.columns:
            ordered_cols.append(col)

wide_df = wide_df[ordered_cols]

wide_df.to_csv("Lung Cancer 2018-2023.csv", index=False)
```
